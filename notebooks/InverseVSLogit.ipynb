# Inverse vs Logit for TopK

This notebook briefly demonstrates the difference between the two topK approaches.

```python
import numpy as np
```

Suppose have 5 places to consider (topK=5) with RAAM costs of:

```python
costs = np.asarray([1.2, 1.3, 1.4, 1.5, 2])
```

Let's calculate the inverse approach which is the inverse divided by the sum of the inverses:

```python
inv = 1 / costs
inv = inv / sum(inv)
inv
```

Now let's calculate the logit-based approach:

```python
log = np.exp(-costs)  # using negative costs as "utility"
log = log / sum(log)
log
```

You'll notice that the approaches are very similar for these cost values. The Logit approach makes the lowest cost values slightly more likely and the higher cost options marginally less likely.
