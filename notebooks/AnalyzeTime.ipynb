# Analyzing Time

This notebook briefly analyzes the time required for the AgentAccess model across different `popPerAgent` parameter settings.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../outputs/Timing.csv")
df["Mins"] = df["Time"] / 60.0
print(len(df))
df.head()
```

```python
x = sorted(set(df["PopPerAgent"]))
x
```

```python
y = df.groupby("PopPerAgent")["Mins"].mean().to_list()
y
```

```python
err = df.groupby("PopPerAgent")["Mins"].std().to_list()
err
```

```python
fig, ax = plt.subplots()
ax.errorbar(x, y, err)
ax.set_yscale("log", nonpositive='mask')
```
